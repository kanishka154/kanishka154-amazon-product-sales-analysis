## 12. Correlation Analysis

Correlation analysis is used to examine the relationships between
key product metrics, including price, discount, rating, reviews,
and estimated monthly purchases.

12.1 Create Correlation Dataset

```python
correlation_df = df[
    [
        "rating",
        "review_count_clean",
        "monthly_purchase_estimate",
        "current_price_clean",
        "Discount %"
    ]
].copy()

correlation_df.head()
```

12.2 Correlation Matrix

```python
correlation_matrix = correlation_df.corr()

correlation_matrix.round(2)
```

12.3 Correlation Heatmap

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    linewidths=0.5,
    center=0
)

plt.title(
    "Correlation Between Product Metrics",
    fontsize=16,
    fontweight="bold",
    pad=15
)

plt.tight_layout()
plt.show()
```

12.4 Price vs Reviews

```python
price_reviews_corr= df[
    ["current_price_clean", "review_count_clean"]
].corr().iloc[0,1]

print(
    "Correlation between Price and Reviews:", price_reviews_corr
)
```

12.5 Price vs Monthly Purchases

```python
price_purchase_corr= df[
    ["current_price_clean", "monthly_purchase_estimate"]
].corr().iloc[0,1]

print(
    "Correlation between Price and Monthly Purchases:",
    round(price_purchase_corr, 2)
)
```

12.6 Discount vs Monthly Purchases

```python
discount_purchase_corr= df[
    ["Discount %", "monthly_purchase_estimate"]
].corr().iloc[0,1]

print(
    "Correlation Between Price and Monthly Purchases:",
    round(discount_purchase_corr,2)
)
```

12.7 Reviews vs Monthly Purchases

```python
reviews_purchase_corr= df[
    ["review_count_clean", "monthly_purchase_estimate"]
].corr().iloc[0,1]

print(
    "Correlation between Reviews and Monthly Purchases:",
    round(reviews_purchase_corr, 2)
)
```

12.8 Correlation Summary

```python
correlation_summary = pd.DataFrame({
    "Relationship": [
        "Price vs Reviews",
        "Price vs Monthly Purchases",
        "Discount vs Monthly Purchases",
        "Reviews vs Monthly Purchases"
    ],
    "Correlation": [
        price_reviews_corr,
        price_purchase_corr,
        discount_purchase_corr,
        reviews_purchase_corr
    ]
})

correlation_summary["Correlation"] = correlation_summary["Correlation"].round(2)

correlation_summary
```

### Customer Engagement Analysis - Business Insights

- Price vs Reviews shows a very weak negative correlation of -0.09, indicating that product price has little linear relationship with review volume.
- Discount vs Monthly Purchases shows a very weak negative correlation of -0.02, suggesting that discount percentage has almost no linear relationship with estimated monthly purchases.
- Reviews vs Monthly Purchases shows a moderate positive correlation of 0.31, indicating that products with more reviews tend to have higher estimated monthly purchases.
- Overall, customer engagement appears to have a stronger relationship with product demand than pricing or discount percentage.
- Correlation measures association and should not be interpreted as proof of causation.

discounts, and marketplace visibility—rather than relying on a single KPI.

## 14. Final Recommendations

Based on the analysis of product performance, pricing, discounts,
customer engagement, and marketplace attributes, the following
recommendations can be considered.

## 1. Focus on High-Engagement Products

Products with strong review volume and estimated monthly purchases
should receive greater attention because they demonstrate stronger
customer engagement and demand.

## 2. Evaluate Discounts Carefully

The relationship between discount percentage and estimated monthly
purchases is very weak. Sellers should therefore avoid assuming that
larger discounts will automatically generate higher demand.

## 3. Use Data-Driven Pricing

The weak relationship between price and review volume suggests that
pricing should be evaluated alongside product quality, customer
engagement, competition, and marketplace positioning rather than
based on price alone.

## 4. Strengthen Customer Engagement

Since reviews and estimated monthly purchases show a positive
relationship, improving customer experience and encouraging genuine
customer feedback may support stronger long-term product engagement.

## 5. Monitor Marketplace Factors

Best Seller status, Sponsored placement, Coupons, Buy Box availability,
and Sustainability Badges should be monitored as additional indicators
of marketplace visibility and product positioning.

## 6. Use Multiple Performance Metrics

Product performance should not be judged using a single metric.
A combination of ratings, reviews, estimated monthly purchases,
price, discount, and marketplace attributes provides a more complete
view of product performance.

### Final Takeaway

The analysis indicates that customer engagement and product demand
are more closely related than price or discount levels alone.
Businesses should therefore combine customer engagement metrics,
pricing strategy, promotional activity, and marketplace visibility
when evaluating product performance.
